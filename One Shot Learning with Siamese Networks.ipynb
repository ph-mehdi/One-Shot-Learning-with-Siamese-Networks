# One-shot face matching

This notebook runs the tested implementation in `face_siamese.py`. Supply an AT&T-style dataset with one directory per person, as shown in the [README](README.md). Training and validation identities are disjoint.


```python
from pathlib import Path
from face_siamese import main

data_dir = Path("data/att_faces")
if not data_dir.is_dir():
    raise FileNotFoundError(f"Add identity folders under {data_dir}; see README.md")
```

## Contrastive training

The model learns from positive and negative image pairs. The training command saves the best development-loss checkpoint.


```python
main(["--data", str(data_dir), "--loss", "contrastive", "--epochs", "10",
      "--checkpoint", "checkpoints/contrastive.pt"])
```

## Triplet training

The same encoder can be trained with anchor, positive, and negative images. This is a separate training run.


```python
main(["--data", str(data_dir), "--loss", "triplet", "--epochs", "10",
      "--checkpoint", "checkpoints/triplet.pt"])
```

The printed values are development losses, not face-recognition accuracy. See the README for evaluation limits.
